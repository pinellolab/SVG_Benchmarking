```python
import warnings
warnings.filterwarnings('ignore')
import os
import pandas as pd
import numpy as np
import scanpy as sc
import squidpy as sq
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage
```

```python
os.makedirs("../../results/03_eval_simulate_data/04_spatial_pattern_clustering", exist_ok=True)
```

```python
def get_gene_correlation(adata):
    filtered_gene_names = [gene for gene in adata.var.index if '_1' in gene]

    # Subset 'adata' to keep only the filtered genes
    adata_filtered = adata[:, filtered_gene_names]
    
    # Calculate the correlation matrix from the gene expression data
    # Transpose the data to get genes as columns
    expr_matrix = adata_filtered.X.T if isinstance(adata_filtered.X, np.ndarray) else adata_filtered.X.toarray().T
    corr_matrix = np.corrcoef(expr_matrix)

    # Convert the correlation matrix to a DataFrame for better handling
    genes = adata_filtered.var.index.tolist()
    corr_df = pd.DataFrame(corr_matrix, index=genes, columns=genes)
    
    return corr_df
```
